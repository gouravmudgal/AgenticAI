```python
from flask import Flask, request, jsonify

app = Flask(__name__)

@app.route("/")
def home():
    return """
    <!DOCTYPE html>
    <html>
    <head>
        <title>Colab Chatbot</title>
        <style>
            body { font-family: Arial; background:#f4f4f4; display:flex; justify-content:center; }
            #chatbox { background:white; width:400px; margin-top:40px; padding:15px; border-radius:8px; }
            #messages { height:300px; overflow-y:auto; border:1px solid #ccc; padding:10px; }
        </style>
    </head>
    <body>
        <div id="chatbox">
            <h2>Chatbot</h2>
            <div id="messages"></div>
            <input id="input" placeholder="Type message..." />
            <button onclick="send()">Send</button>
        </div>

        <h3>Chat History</h3>
        <div id="chat-history"></div>

        <script>
            function send() {
                let input = document.getElementById("input");
                let msg = input.value;
                const chatHistory = document.getElementById('chat-history');
                if (!msg) return;
                chatHistory.innerHTML += "<div><strong>You:</strong> " + msg + "</div>";
                console.log('You: '+ msg)

                messages.innerHTML += `<div><b>You:</b> ${msg}</div>`;
                input.value = "";

                fetch("/chat", {
                    method: "POST",
                    headers: {"Content-Type": "application/json"},
                  body: JSON.stringify({message: msg})
                })
                .then(r => r.json())
                .then(d => {
                    messages.innerHTML += `<div><b>Response:</b> ${d.reply}</div>`;
                    chatHistory.innerHTML += "<div><strong>Response:</strong> " + d.reply + "</div>";
                    console.log('Response: '+ d.reply)
                });

            }
        </script>

    </body>
    </html>
    """

@app.route("/chat", methods=["POST"])
def chat():
    msg = request.json.get("message", "").lower()
    model = "gemini-2.5-flash"
    reply = client.models.generate_content(
    model=model,
    contents=msg
)

    return jsonify({"reply": reply.text})
```
